## Setup

`psycopg` is the Postgres driver (the standard raw driver for the Python).
`python-dotenv` loads `DATABASE_URL` out of a `.env` file, so the password never lives in
your code. then we point `DATABASE_URL` at a real local Postgres server so
every query below runs for real, not simulated.

## Task 0 - Run Postgres, one command
*"Before your app can talk to a database, the database has to exist."*


this stage is one command:

```bash
docker run --name taskdb -e POSTGRES_PASSWORD=dev -e POSTGRES_DB=tasks -p 5432:5432 -d postgres
```

That pulls the official `postgres` image (if you don't have it yet) and starts it as a
container named `taskdb`, with a database called `tasks` already created, listening on
port 5432.

**Checkpoint:** `docker ps` shows `taskdb` as `Up`; `docker exec -it taskdb psql -U postgres -d tasks` connects you to a `psql` prompt.
